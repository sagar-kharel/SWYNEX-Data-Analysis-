```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as pt
import seaborn as sns
```

```python
# all data import
customers= pd.read_csv(r"D:\Python Data\relational dataset\customers.csv")
orders = pd.read_csv(r"D:\Python Data\relational dataset\orders.csv")
payments = pd.read_csv(r"D:\Python Data\relational dataset\payments.csv")
products = pd.read_csv(r"D:\Python Data\relational dataset\products.csv")
```

```python
#viewed all columns of all files to see common columns
print(customers.columns,
orders.columns,
payments.columns,products.columns)
```

```python
#combining to make single file
df = orders.merge(customers,on='CustomerID', how="left")
```

```python
df = df.merge(products,on='ProductID',how="left")
```

```python
df = df.merge(payments,on='OrderID',how="left")
```

```python
#Identifing missing values
total_null=df.isnull().sum()
(total_null>1).sum()
only_null= total_null[total_null>1].sort_values(ascending=False)
only_null
```

```python
numeric= df.select_dtypes(include='number').columns
categorical = df.select_dtypes(include='object').columns
print("numeric=",(numeric),"\n\n\ncategorical=",(categorical))
```

```python
# Filling numerical nulls
df["Quantity"] = df["Quantity"].fillna(
    df["Quantity"].median()
)
```

```python
df["Discount"] = df["Discount"].fillna(
    df["Discount"].median()
)
```

```python
df["Age"] = df["Age"].fillna(
    df["Age"].median()
)
```

```python
df.isnull().sum().sort_values(ascending=False)
```

```python
# Filling categorical nulls
df["City"] = df["City"].fillna('unknown')
df["PaymentMethod"] = df["PaymentMethod"].fillna('unknown')
df["OrderDate"] = df["OrderDate"].fillna('unknown')
df["PaymentDate"] = df["PaymentDate"].fillna('unknown')
df["SignupDate"] = df["SignupDate"].fillna('unknown')
df["CustomerSegment"] = df["City"].fillna('CustomerSegment')
```

```python
df["Discount"] = df["Discount"].fillna(0)
```
